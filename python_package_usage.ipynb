# city-macro-data Python 使用教程

这个 notebook 演示如何使用 `city-macro-data` Python package：安装、读取、校验、分析与更新流程。

## 1) 环境准备

如果你在项目根目录，建议先安装包：

```bash
python -m pip install .\\python-pkg
```

如果你只是使用者（不是开发者），可以安装发布后的 wheel 或私有源包。

````python
python -m pip install .\\python-pkg
````

````python
from city_macro_data import load_data, get_metadata, data_version, validate_data
````

## 2) 读取数据与元数据

````python
df = load_data()
meta = get_metadata()

print("shape:", df.shape)
print("data_version:", data_version())
print("metadata keys:", list(meta.keys()))
df.head()
````

## 3) 数据质量校验

`validate_data()` 默认检查：
- 数据非空
- 必需列存在（默认 `年份`、`城市`）

````python
validate_data()
````

````python
# 自定义必需列
validate_data(["年份", "城市", "省份"])
````

## 4) 常见分析示例

````python
# 例1：按年份统计城市数量
city_count_by_year = df.groupby("年份")["城市"].nunique()
city_count_by_year.head()
````

````python
# 例2：查看缺失值最多的前10列
missing_top10 = df.isna().sum().sort_values(ascending=False).head(10)
missing_top10
````

## 5) 数据更新后的标准流程（开发者）

当你替换了 `data-src/city_macro.xlsx` 后，按顺序执行：

```bash
python build-data/build_data.py
powershell -ExecutionPolicy Bypass -File scripts/sync_packages.ps1
python -m pip install --force-reinstall .\\python-pkg
```

然后回到本 notebook 重新运行读取和校验单元。

## 6) 常见报错排查

- `ModuleNotFoundError: No module named city_macro_data`：包未安装，先执行安装命令。
- `UnicodeEncodeError`（Windows, editable 模式）：路径包含特殊字符时尽量使用非 `-e` 安装或英文路径。
- `openpyxl` / `numpy` 兼容报错：升级 `openpyxl` 到 `>=3.1.5`。
