# citymacrodata R 使用教程

这个 notebook 演示如何使用 `citymacrodata` R package：安装、读取、校验、分析与更新流程。

## 1) 环境准备

在项目根目录可以这样安装本地 R 包：

```bash
R CMD INSTALL r-pkg
```

若要正式分发，建议先 `R CMD build r-pkg` 再安装 tar.gz。

````r
library(citymacrodata)
````

## 2) 读取数据与元数据

````r
df <- load_data()
meta <- get_metadata()

print(dim(df))
print(data_version())
print(names(meta))
head(df)
````

## 3) 数据质量校验

`validate_data()` 默认检查：
- 数据非空
- 必需列存在（默认 `年份`、`城市`）

````r
validate_data()
````

````r
# 自定义必需列
validate_data(c("年份", "城市", "省份"))
````

## 4) 常见分析示例

````r
# 例1：按年份统计城市数量
city_count_by_year <- aggregate(城市 ~ 年份, data = df, FUN = function(x) length(unique(x)))
head(city_count_by_year)
````

````r
# 例2：每列缺失值个数（前10）
na_counts <- sort(colSums(is.na(df)), decreasing = TRUE)
head(na_counts, 10)
````

## 5) 数据更新后的标准流程（开发者）

当替换了 `data-src/city_macro.xlsx`，请按顺序执行：

```bash
python build-data/build_data.py
powershell -ExecutionPolicy Bypass -File scripts/sync_packages.ps1
R CMD INSTALL r-pkg
```

然后重新运行本 notebook 的读取和校验单元。

## 6) 常见报错排查

- `there is no package called 'citymacrodata'`：先安装本地包。
- `could not find function validate_data`：包版本过旧，重新安装 `r-pkg`。
- 读取结果仍是旧数据：通常是漏跑了 `build_data.py` 或 `sync_packages.ps1`。
